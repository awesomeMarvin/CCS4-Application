# New Section

For this section, I used a match-case statement to determine the input of the user. If their input is among the given options, it would execute the appropriate code block, otherwise, the entire function would get called again.

```python
def BootUp():
    while True:
        print("[Marvin's Automatic Teller Machine]")
        choice1 = str(input("[1] Log in\n"
                            "[2] Create Account\n"
                            "[3] Shut Down\n"
                            ": "))
        match choice1:
            case "1":
                lib.space()
                l = lib.Account.Login()
                if l == 0:
                    continue
                return
            case "2":
                lib.space()
                lib.AccountCreation()
                continue
            case "3":
                lib.space()
                print("Shutting down...")
                sys.exit()
            case "101":
                lib.space()
                lib.Account.Test()
            case _:
                lib.space()
                print("Error! Please choose among the given options.")
                continue
```

For this section below, it would only get executed after a successful log in.

```python
def Menu():
    while True:
        print("[Hi! What would you like to do today?]")
        choice2 = str(input("[1] Check Balance \n"
                            "[2] Deposit \n"
                            "[3] Withdraw \n"
                            "[4] Transaction History \n"
                            "[5] Log Out"
                            ": "))

        match choice2:
            case "1":
                lib.space()
                lib.Account.ViewBalance()
                continue
            case "2":
                lib.space()
                lib.Account.Deposit()
                continue
            case "3":
                lib.space()
                lib.Account.Withdraw()
                continue
            case "4":
                lib.space()
                lib.Account.TransactionHistory()
                continue
            case "5":
                lib.space()
                print("Logging out..\n")
                BootUp()
            case _:
                lib.space()
                print("Error! Please choose among the given choices!\n")
                continue
```

After setting those up, it's time to run the machine

```python
lib.space()
BootUp()
Menu()
```
